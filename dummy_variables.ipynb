```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
dfle = df
dfle = dfle.drop(columns=['Property ID'], errors='ignore')
dfle["Town"] = le.fit_transform(dfle["Town"])
dfle["Property Type"] = le.fit_transform(dfle["Property Type"])
dfle
```

```python
X = dfle[['Square Feet', 'Bedrooms', 'Bathrooms', 'Town', 'Property Type']]
X
```

```python
y = dfle['Price ($)']
y
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LinearRegression
import pandas as pd

X = dfle[['Square Feet', 'Bedrooms', 'Bathrooms', 'Town', 'Property Type']]
y = dfle['Price ($)']

preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(), [3, 4])
    ],
    remainder='passthrough'
)

model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', LinearRegression())
])
```

```python
model.fit(X, y)

# 5. Make a prediction (passing a DataFrame with the same original column names)
sample_house = pd.DataFrame([[2400, 3, 2, 0, 1]], 
                            columns=['Square Feet', 'Bedrooms', 'Bathrooms', 'Town', 'Property Type'])

predicted_price = model.predict(sample_house)
print("Predicted House Price:", predicted_price)
```
